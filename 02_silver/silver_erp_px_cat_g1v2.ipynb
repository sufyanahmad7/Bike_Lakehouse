### Silver transformations

Rename columns with proper column names.

```python
RENAME_MAP = {
    "ID": "category_id",
    "CAT": "category",
    "SUBCAT": "sub_category",
    "MAINTENANCE": "maintenances",
}

for old_name, new_name in RENAME_MAP.items():
    df = df.withColumnRenamed(old_name, new_name)
```

### Write into Silver table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.silver.erp_parts_categories")
```

```python
%sql
SELECT * 
FROM workspace.silver.erp_parts_categories
LIMIT 10
```
