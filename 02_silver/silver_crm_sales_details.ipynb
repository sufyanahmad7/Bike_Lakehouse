### Silver `transformations`

Rename columns with proper column names

```python
RENAME_MAP = {
    "sls_ord_num": "sales_order_num",
    "sls_prd_key": "sales_product_key",
    "sls_cust_id": "sales_customer_id",
    "sls_order_dt": "sales_order_dateproduct_cost",
    "sls_ship_dt": "sales_shipping_date",
    "sls_due_dt": "sales_due_date",
    "sls_sales": "sales_sales",
    "sls_quantity": "sales_quality",
    "sls_price": "sales_price",
}

for old_name, new_name in RENAME_MAP.items():
    df = df.withColumnRenamed(old_name, new_name)
```

### Write into Silver table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.silver.crm_sales")
```
