### Gold transformations & write into Gold table

Join tables on product_id

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.gold;

CREATE OR REPLACE TABLE workspace.gold.dim_product
USING DELTA
AS
SELECT
    a.product_id,
    a.product_key,
    a.product_key_simplified,
    a.product_name,
    a.product_cost,
    a.product_line,
    a.product_start_date,
    a.product_end_date,
    b.category,
    b.sub_category,
    b.maintenance
    FROM workspace.silver.crm_products AS a
LEFT JOIN workspace.silver.erp_parts_categories AS b
    ON a.product_key_simplified = b.category_id
```
