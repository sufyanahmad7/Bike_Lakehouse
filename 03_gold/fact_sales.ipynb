### Gold transformations & write into Gold table

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.gold;

CREATE OR REPLACE TABLE workspace.gold.fact_sales
USING DELTA
AS
SELECT
    sales_order_num,
    sales_product_key,
    sales_customer_id,
    sales_order_date,
    sales_shipping_date,
    sales_due_date,
    sales_total_amount,
    sales_price
    sales_quality
FROM workspace.silver.crm_sales
```
